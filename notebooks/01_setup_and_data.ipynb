```python
!pip install pandas
import urllib.request
import json
import os
import pandas as pd

# 1. Define URLs and file paths
halueval_url = "https://raw.githubusercontent.com/RUCAIBox/HaluEval/main/data/qa_data.json"
raw_data_path = "./data/qa_data_raw.json"
processed_data_path = "./data/prototype_qa_data.csv"

print("Downloading HaluEval QA dataset...")

# 2. Download the raw data if it doesn't exist yet
if not os.path.exists(raw_data_path):
    urllib.request.urlretrieve(halueval_url, raw_data_path)
    print("Download complete.")
else:
    print("Raw data already exists.")

# 3. Load the JSON data
with open(raw_data_path, 'r', encoding='utf-8') as f:
    raw_data = [json.loads(line) for line in f]

# 4. Convert to a Pandas DataFrame for easy viewing and manipulation
df = pd.DataFrame(raw_data)

# 5. Extract a small subset for the prototype (first 500 records)
# We keep only the columns we need for now: knowledge, question, right_answer
df_subset = df[['knowledge', 'question', 'right_answer']].head(500)

# 6. Save the cleaned subset to the data folder
df_subset.to_csv(processed_data_path, index=False)

print(f"\nSaved {len(df_subset)} records to {processed_data_path}")

# 7. Display the first 3 rows to inspect the structure
df_subset.head(3)
```
