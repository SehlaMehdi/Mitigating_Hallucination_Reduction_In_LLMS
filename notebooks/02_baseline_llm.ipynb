```python
# 1. Install the groq python client
!pip install groq -q

import os
import pandas as pd
from groq import Groq
from google.colab import userdata
from getpass import getpass

# 2. Securely input your Groq API key
print("Please enter your Groq API Key:")
os.environ["GROQ_API_KEY"] = getpass()

# 3. Initialize the Groq Client
client = Groq()

# 4. Define SYSTEM A (Baseline LLM without RAG)
def system_a_baseline(question):
    """
    Sends the user question directly to Llama via Groq.
    No external context is provided.
    """
    try:
        response = client.chat.completions.create(
            # Using Meta's fast and highly capable 8B model on Groq
            model="llama-3.1-8b-instant",
            messages=[
                {
                    "role": "system",
                    "content": "You are a helpful AI assistant. Answer the user's question directly and concisely."
                },
                {
                    "role": "user",
                    "content": question
                }
            ],
            temperature=0.0, # Keep temperature at 0 for factual, reproducible answers
            max_tokens=200
        )
        return response.choices[0].message.content.strip()
    except Exception as e:
        return f"Error: {str(e)}"

# 5. Let's test System A on the first question from our dataset
print("\nLoading dataset to test System A...")
df = pd.read_csv("./data/prototype_qa_data.csv")

# Get the very first question and its known right answer
test_question = df.iloc[0]['question']
ground_truth = df.iloc[0]['right_answer']

print(f"QUESTION: {test_question}")
print(f"\nGROUND TRUTH (From HaluEval): {ground_truth}")

# Run System A
baseline_answer = system_a_baseline(test_question)
print(f"\nSYSTEM A (Baseline) ANSWER:\n{baseline_answer}")
```
